```python
import numpy as np

runs = np.array([45, 67, 12, 89, 34])
balls = np.array([30, 50, 15, 60, 28])

strike_rate = runs / balls * 100
print(np.round(strike_rate, 2))
```

```python
print("Total runs:", runs.sum())
print("Average:", runs.mean())
print("Median:", np.median(runs))
print("Highest:", runs.max())
print("Lowest:", runs.min())
print("Std deviation:", runs.std())
print("Index of highest score:", runs.argmax())
```

```python
fours = np.array([5, 8, 1, 10, 3])
sixes = np.array([1, 2, 0, 4, 1])

#1. Boundary runs in each innings (a four is worth 4 runs and a six is worth 6)
boundary_runs = fours * 4 + sixes * 6
print(boundary_runs)

#2. Percentage of runs from boundaries in each innings, rounded to 1 decimal place
runs_perc = (boundary_runs / runs) * 100
print(np.round(runs_perc, 1))

#3. Total sixes across all innings
total_sixes = sixes.sum()
print(total_sixes)

#4. Which innings (by index) had the most fours
print(fours.argmax())

#5. Average strike rate across the 5 innings, rounded to 2 decimal places
total_runs = runs.sum()
total_balls = balls.sum()
avgStr = (total_runs / total_balls) * 100
print(total_runs)
print(total_balls)
print(np.round(avgStr,2))
```
