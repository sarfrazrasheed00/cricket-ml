# Lesson 2.1: Pandas, Series aur DataFrame

## Pandas kyun?
NumPy ke do masle:
1. Columns ke **naam nahi**: yaad rakhna parta tha ke `X[:, 1]` fours hai
2. **Text** (player, opposition) achha nahi chalta

| | NumPy | Pandas |
|---|---|---|
| Column ke naam | ❌ `X[:, 1]` | ✅ `df["fours"]` |
| Numbers + text ek saath | ❌ | ✅ |
| CSV, missing values, grouping | mushkil | ek line |

Do cheezein yaad rakhein:
- **DataFrame** = poora table (MySQL table jaisa)
- **Series** = us table ka **ek column**

Andar se Pandas, NumPy hi use karta hai, to vectorisation aur mask sab yahan bhi chalte hain.

````python
import pandas as pd
````

## Part 1: DataFrame banana

Dictionary: har **key = column ka naam**, har **value = us column ki list**. Saari lists ki length barabar (8 innings).

Aakhri line pe sirf `df` likha hai, `print(df)` nahi. Notebook **cell ki aakhri line** khud dikhati hai, aur DataFrame **saaf table** ki shakal mein aata hai.

````python
data = {
    "player":           ["Babar", "Rizwan", "Babar", "Fakhar", "Rizwan", "Babar", "Fakhar", "Rizwan"],
    "opposition":       ["India", "Australia", "England", "India", "England", "Australia", "England", "India"],
    "batting_position": [3, 1, 3, 2, 1, 3, 2, 1],
    "balls":            [30, 50, 15, 60, 28, 3, 70, 40],
    "fours":            [5, 8, 1, 10, 3, 0, 12, 4],
    "sixes":            [1, 2, 0, 4, 1, 0, 3, 2],
    "runs":             [45, 67, 12, 89, 34, 0, 102, 51],
}

df = pd.DataFrame(data)
df
````

### `describe()` se 4 professional sawal
1. **`count` sab columns mein barabar?** Kam ho to **missing values** hain
2. **`min`/`max` sense banate hain?** −5 balls ya 900 runs = ghalti (**sanity check**, cricket knowledge)
3. **Mean aur median mein kitna farq?** Bohat zyada = bare scores (outliers) average kheench rahe hain
4. **Kya ye statistic us column ke liye matlab rakhti hai?** `batting_position` ka mean **2.0**: "average position 2" ka koi matlab nahi (Part 6 mein)

Text column ka describe alag hota hai (`count`, `unique`, `top`, `freq`):

````python
df["player"].describe()
````

**std ka chhota farq:** Pandas ka `std` (35.33) **sample** std hai (n − 1 se divide), NumPy ka `.std()` **population** std (n se). Numbers thore alag aayein to pareshan na hon.

## Part 3: Columns select karna

| Code | Result | Shape |
|---|---|---|
| `df["runs"]` | **Series** (ek column) | `(8,)` 1D |
| `df[["runs"]]` | **DataFrame** (ek column wala table) | `(8, 1)` 2D |
| `df[["player", "runs"]]` | **DataFrame** (do columns) | `(8, 2)` |

**Single bracket = 1D, double bracket = 2D.** Bilkul NumPy wala farq.

````python
print(df["runs"])
print(type(df["runs"]))
````

````python
print(df["runs"].shape)     # (8,)   → 1D
print(df[["runs"]].shape)   # (8, 1) → 2D
````

````python
df[["player", "runs"]]
````

````python
print(df["runs"].mean())
print(df["sixes"].max())
````

### `batting_position`: number hai, lekin asal mein **categorical (ordinal)**

- No. 3, opener (No. 1) se **"teen guna"** nahi hota. Ye numbers sirf **label** hain

### 3. Mean vs median

````python
print(f"Runs Mean:   {df['runs'].mean()}")
print(f"Runs Median: {df['runs'].median()}")
````

Mean (50) median (48) se zyada kyunki **102** mean ko upar kheenchta hai. Median sirf beech (45, 51) dekhta hai.
Agar 102 ki jagah 200 hota: median **phir bhi 48**, mean **62**. Isi liye outliers ho to **median** zyada bharosemand (Ames mein median imputation isi wajah se).

### 4. Naye features: boundary_runs aur boundary_pct

````python
df["boundary_runs"] = df["fours"] * 4 + df["sixes"] * 6
df["boundary_pct"] = df["boundary_runs"] / df["runs"] * 100
df
````

Index 5 (Babar ka **duck**): runs = 0, boundary = 0 → `0 / 0` = **NaN**.
- Normal Python mein `0/0` = `ZeroDivisionError`. Pandas **error nahi deta**, chupke se NaN rakh deta hai
- **Feature engineering khud bhi missing values paida kar sakti hai**

````python
df["boundary_pct"].isnull().sum()   # 1
````

(Isay theek karna Lesson 2.2 mein: `df.loc[df["runs"] == 0, "boundary_pct"] = 0`)

## Part 7: Data leakage (sab se important sabaq)

Har ML project ka pehla sawal:
> **"Prediction KIS WAQT ho rahi hai?"**

Agar **innings se pehle** predict karna hai: `balls`, `fours`, `sixes` sab innings **ke baad** pata chalte hain.
Model asal mein ye seekhega:
```
runs ≈ 4 × fours + 6 × sixes + (baaqi singles/doubles)
```
Yaani **jawab ka hissa pehle se dekh ke** jawab. Training/testing mein accuracy zabardast, lekin asal prediction ke waqt ye values **hain hi nahi** → model bekaar.

| Prediction kab? | Theek features | Leakage |
|---|---|---|
| **Innings se pehle** | batting position, opposition, venue, recent form, career average | balls, fours, sixes, strike_rate, out kaise hua |
| **Innings ke beech** (10 overs baad) | upar wale + ab tak ki balls, runs, wickets, run rate | final balls, final fours |

> Jab koi feature target se **"bohat achha"** jura ho, khush hone se pehle poochein: **ye value KAB pata chalti hai?**
