```python
# 2. Data types
print(df.dtypes)

# Numerical:   balls, fours, sixes, runs, strike_rate
# Categorical: player, opposition
#
# batting_position int64 mein store hai, lekin asal mein CATEGORICAL (ordinal) hai.
# "Average batting position 2.0" ka koi matlab nahi; ye numbers sirf labels hain.
# (Har position pe player ke average RUNS nikalna meaningful hai, wo groupby se karenge.)
#
# Sabaq: dtype batata hai data KAISE store hua hai,
#        column ASAL mein kya hai, ye domain knowledge decide karta hai.
```

```python
# 3. Mean vs median
print(f"Runs Mean:   {df['runs'].mean()}")
print(f"Runs Median: {df['runs'].median()}")

# Mean (50) median (48) se zyada hai kyunki 102 jaisa bara score
# mean ko upar kheench leta hai. Median sirf beech ki values (45, 51) dekhta hai,
# isliye outliers ka us pe asar nahi hota.
```

```python
# 4. Naye features
df["boundary_runs"] = df["fours"] * 4 + df["sixes"] * 6
df["boundary_pct"] = df["boundary_runs"] / df["runs"] * 100

# Index 5 (duck): runs = 0, to 0/0 = NaN (missing value).
# Pandas error nahi deta, chupke se NaN rakh deta hai.
# Feature engineering khud bhi missing values paida kar sakti hai.
df
```
