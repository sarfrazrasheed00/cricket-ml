```python
import numpy as np

runs  = np.array([45, 67, 12, 89, 34, 0, 102, 51])
balls = np.array([30, 50, 15, 60, 28, 3, 70, 40])

print(runs[4])

recent_form = runs[-4:].sum() / 4
print(recent_form)

duck_count = (runs == 0).sum() 
print(duck_count)

fifty_plus_runs = runs[runs >= 50]
fifty_plus_balls = balls[runs >= 50]
strike_rate = (fifty_plus_runs / fifty_plus_balls) * 100
print(np.round(strike_rate, 2))

filtered_runs = runs[(balls < 20) | (runs >= 100)]
print(filtered_runs)
```
