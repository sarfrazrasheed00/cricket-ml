```python
import numpy as np

# columns: balls, fours, sixes
X = np.array([
    [30, 5, 1],
    [50, 8, 2],
    [15, 1, 0],
    [60, 10, 4],
    [28, 3, 1],
])

y = np.array([45, 67, 12, 89, 34])   # runs

print(X[0])        # pehli innings (poori row)
print(X[:, 0])     # balls column (saari innings)
print(X[1, 2])     # 2nd innings ke sixes
print(X[:3])       # pehli 3 innings
print(X[:, 1:])    # fours aur sixes columns

print(X.mean(axis=0))   # har feature ka average
print(X.max(axis=0))    # har feature ka max
print(X.sum(axis=1))    # har row ka sum
```

```python
new_innings = np.array([35, 4, 2])
print(new_innings.shape)

new_innings_2d = new_innings.reshape(1, -1)
print(new_innings_2d)
print(new_innings_2d.shape)
```

```python
#1. X.shape use karke print karein: "Innings: 5, Features: 3" (numbers hardcode nahi karne, shape se nikalne hain) 
rows = X.shape[0]
columns = X.shape[1]
message = f"""
Innings: {rows}, Features: {columns}
"""

print(message)

#2. Saari innings ke total fours
total_fours = X[:, 1].sum()
print(total_fours)

#3. Woh innings jinmein 2 ya zyada sixes lage: unki poori rows (X se) aur unke runs (y se) print karein
mask = X[:, 2] >= 2

# Filtering X and y
matching_rows = X[mask]
matching_runs = y[mask]

print("Matching Rows from X:")
print(matching_rows)

print("\nMatching Runs from y:")
print(matching_runs)

#4. Har innings ke boundary runs (fours×4 + sixes×6) X ke columns se nikalein
boundary_runs = (X[:, 1] * 4) + (X[:, 2] * 6)

print("Har inning ke boundary runs:")
print(boundary_runs)

#5. Ek nayi innings banayein (42 balls, 6 fours, 1 six), usay model ke liye tayyar shape mein reshape karein, aur shape print karein
# Nayi inning ka data (1D array)
new_inning = np.array([42, 6, 1])

# Model ke liye reshape karna (2D array: 1 row, 3 columns)
new_inning_reshaped = new_inning.reshape(1, -1)

print("Reshaped Data:")
print(new_inning_reshaped)

print("\nData ki nayi Shape:")
print(new_inning_reshaped.shape)
```

```python
w = np.array([0.8, 2.0, 3.0])
b = 2

y_pred = X @ w + b
error = y - y_pred

mae = np.abs(error).mean()
mse = (error ** 2).mean()

print(f"MAE: {mae}")
print(f"MSE: {mse}")  
```

```python
for b in [2, 3, 4, 4.5, 5, 6]:
    y_pred = X @ w + b
    mse = ((y - y_pred) ** 2).mean()
    print(f"b = {b}  →  MSE = {mse:.2f}")
```
